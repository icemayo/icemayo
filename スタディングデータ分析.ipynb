Seriesの基本操作

 Seriesの作成方法

```python
import pandas as pd
import numpy as np
```

```python
# 1. リストから作成
numbers = pd.Series([1,2,3,4,5])
print("基本的な Series:")
print(numbers)

# 2. インデックスを指定して作成
fruits = pd.Series([100,200,300], index=['りんご','みかん','バナナ'])
print("\n果物の価格 Series:")
print(fruits)

# 3. 辞書から作成
scores = pd.Series({"国語":85,"数学":90, "英語":88})
print("\nテストのスコア Series:")
print(scores)

# 4. NumPy 配列から作成
array_data = np.array([10,20,30,40])
array_series = pd.Series(array_data)
print("\nNumPy 配列から作成した Series:")
print(array_series)

# 5. スカラー値と指定された長さから作成
scalar_series = pd.Series(5, index=['a', 'b', 'c', 'd'])
print("\nスカラー値から作成した Series:")
print(scalar_series)
```

Series の基本的な演算

```python
# Series 間の演算
s1 = pd.Series([1,2,3], index=['a', 'b', 'c'])
s2 = pd.Series([4,5,6], index=['a', 'b', 'c'])

# 加算
print("Series の加算:")
print(s1 + s2)

# 乗算
print("\nSeries の乗算:")
print(s1 * 2)

# 関数適用
print("\n関数適用（平方根）:")
print(np.sqrt(s1))

# インデックスが異なる場合の演算
s3 = pd.Series([7, 8, 9], index=['a', 'b', 'c'])
print("\nインデックスが異なる Series の加算:")
print(s1 + s3) # 共通しないインデックスは NaN になります
```

DataFrame の作成と基本操作

1. 辞書からの作成

最も一般的な方法は、列名をキーとする辞書から DataFrame を作成する方法です。

```python
# 生徒の成績データ
data = {
    '名前': ['田中', '佐藤', '山田', '鈴木', '伊藤'],
    '国語': [85, 90, 88, 75, 93],
    '数学': [92, 85, 95, 80, 88],
    '英語': [90, 88, 85, 95, 82]
}

df = pd.DataFrame(data)
print("生徒の成績 DataFrame:")
print(df)

# インデックスを指定して作成
df_with_index = pd.DataFrame(data, index=['学生1', '学生2', '学生3', '学生4', '学生5'])
print("|nインデックス指定のDataFrame：")
print(df_with_index)
```

2. リストからの作成

2 次元リスト（リストのリスト）からも DataFrame を作成できます。

```python
# 商品の売上データ
sales_data = [
    ['A商品', 100, 150, 200],
    ['B商品', 120, 130, 180],
    ['C商品', 90, 160, 210],
    ['D商品', 110, 140, 190]
]

df_sales = pd.DataFrame(
    sales_data,
    columns=['商品名',  '1月', '2月', '3月']
)
print("\n商品別売上 DataFrame:")
print(df_sales)
```

3. Series の辞書からの作成

複数の Series オブジェクトを組み合わせて DataFrame を作成することもできます。



```python
# 複数の Series からの作成
products = pd.Series(['ノートPC', 'タブレット', 'スマートフォン'])
prices = pd.Series([80000, 60000, 40000])
stock = pd.Series([15, 20, 30])

df_products = pd.DataFrame({
    '商品名': products,
    '価格': prices,
    '在庫数':stock
})

print("\n商品情報 DataFrame:")
print(df_products)
```

4. NumPy 配列からの作成

NumPy の多次元配列を使って DataFrame を作成することも可能です。



```python
# NumPy 配列からの作成
np_data = np.random.randn(3, 4) # 3行4列のランダムな値
df_numpy = pd.DataFrame(
    np_data,
    columns=['A', 'B', 'C', 'D'],
    index=['行1', '行2', '行3']
)
print("\nNumPy 配列から作成した DataFrame:")
print(df_numpy)
```
